## Tips DebuggingKalau error, jangan panik. Pesan error Python punya struktur yang bisa kamu baca:```Traceback (most recent call last):  File "script.py", line 5, in <module>    print(x[10])IndexError: list index out of range```Artinya: error di file `script.py`, baris 5, jenis `IndexError` → kamu mengakses indeks di luar jangkauan. Biasanya path error menunjuk ke baris kode yang **memanggil** operasi, bukan baris yang menyebabkannya — jadi cek baris itu dan baris-baris sebelumnya.Jenis error yang umum di chapter ini:- `IndentationError` — spasi/tab tidak konsisten. Pakai 4 spasi untuk tiap level indentasi.- `NameError` — variabel belum didefinisikan atau salah ketik.- `TypeError` — operasi dengan tipe data yang salah (misal `1 + "2"`).- `IndexError` — akses list dengan indeks di luar range. Indeks mulai dari 0.- `KeyError` — akses dict dengan kunci yang tidak ada.- `ZeroDivisionError` — bagi dengan 0.- `SyntaxError` — salah ketik sintaks (kurung, kutip, titik dua).- `ValueError` — tipe benar tapi nilai tidak valid (misal `int("abc")`).Kalau masih bingung, salin pesan error dan diskusikan di sesi berikutnya. Pesan error adalah informasi paling berharga saat debugging — tanpa itu, masalahnya cuma "kok tidak jalan" tanpa tahu kenapa.
