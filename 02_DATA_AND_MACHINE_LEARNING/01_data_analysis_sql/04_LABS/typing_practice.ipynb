## Soal 1: Array Sederhana

Tujuan: membiasakanmu dengan sintaks dasar NumPy. Buat array 1D berisi bilangan genap dari 0 sampai 20 (inklusif), lalu cetak array-nya. Hint: ada satu fungsi NumPy yang langsung bisa menghasilkan deret angka dengan step tertentu.

**Referensi singkat untuk Soal 1**:

```python
import numpy as np
arr = np.arange(0, 21, 2)
print(arr)
```

`np.arange(start, stop, step)` menghasilkan array dengan nilai dari `start` sampai `stop-1` dengan `step` tertentu. Karena 20 harus inklusif, kita tulis `stop=21` (bukan 20).

## Soal 2: Indexing dan Slicing

Buat array 1D berisi angka 0 sampai 9. Dari array itu, cetak: (a) elemen ke-3 (ingat, indexing mulai dari 0), (b) 4 elemen pertama, (c) 3 elemen terakhir, (d) array dalam urutan terbalik.

**Referensi singkat untuk Soal 2**:

```python
import numpy as np
arr = np.arange(10)
print(arr[2])      # elemen ke-3 (index 2)
print(arr[:4])     # 4 elemen pertama
print(arr[-3:])    # 3 elemen terakhir
print(arr[::-1])   # reverse
```

Slicing `arr[-3:]` berarti "dari index -3 sampai akhir". Slicing `arr[::-1]` adalah trik umum untuk reverse: `start` dan `stop` kosong, `step=-1`.

## Soal 3: Matriks 2D

Buat matriks 3×3 berisi angka 1 sampai 9 (hint: `np.arange(1, 10).reshape(3, 3)`). Cetak: (a) elemen di baris ke-2 kolom ke-2 (yaitu angka 5), (b) seluruh baris pertama, (c) seluruh kolom terakhir.

**Referensi singkat untuk Soal 3**:

```python
import numpy as np
mat = np.arange(1, 10).reshape(3, 3)
print(mat[1, 1])   # baris 1, kolom 1 (zero-indexed)
print(mat[0, :])   # seluruh baris 0
print(mat[:, -1])  # seluruh kolom terakhir
```

Format indexing 2D di NumPy: `mat[baris, kolom]`. Tanda `:` artinya "semua index di sumbu itu".

## Soal 4: Boolean Indexing

Buat array berisi 20 angka acak 0 sampai 100. Pakai boolean indexing untuk mengambil hanya angka yang **genap** dan **lebih besar dari 50**. Cetak array hasil filter dan juga hitung ada berapa elemen.

**Referensi singkat untuk Soal 4**:

```python
import numpy as np
arr = np.random.randint(0, 101, 20)
mask = (arr % 2 == 0) & (arr > 50)
filtered = arr[mask]
print(f"Asli    : {arr}")
print(f"Filter  : {filtered}")
print(f"Jumlah  : {len(filtered)}")
```

Boolean indexing: buat mask (array boolean), lalu pakai mask itu untuk menyaring array asli. Untuk AND multi-kondisi, gunakan `&` (bukan `and`).

## Soal 5: Operasi Vektor

Buat dua array `a = [10, 20, 30, 40, 50]` dan `b = [1, 2, 3, 4, 5]`. Hitung dan cetak: (a) `a + b`, (b) `a * b`, (c) `a / b`, (d) `a ** 2`, (e) `a - 5`.

**Referensi singkat untuk Soal 5**:

```python
import numpy as np
a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])
print(a + b)
print(a * b)
print(a / b)
print(a ** 2)
print(a - 5)
```

Semua operasi di atas adalah element-wise. Beda dengan list Python: kalau `a + b` di list adalah concatenation, di NumPy array adalah penjumlahan elemen per elemen.

## Soal 6: Agregasi dan Statistik

Buat array berisi 50 angka acak 60-100. Hitung dan cetak: (a) mean, (b) median, (c) standar deviasi, (d) nilai minimum dan maksimum, (e) index dari nilai maksimum.

**Referensi singkat untuk Soal 6**:

```python
import numpy as np
nilai = np.random.randint(60, 101, 50)
print(f"Mean : {nilai.mean():.2f}")
print(f"Median: {np.median(nilai):.2f}")
print(f"Std  : {nilai.std():.2f}")
print(f"Min  : {nilai.min()}")
print(f"Max  : {nilai.max()}")
print(f"Idx max: {nilai.argmax()}")
```

Perhatikan: `.mean()`, `.max()`, `.min()` adalah method, sedangkan `np.median()` adalah fungsi. Keduanya menghasilkan hasil yang sama untuk operasi yang didukung keduanya, tapi method lebih idiomatik untuk chaining.

## Soal 7: Broadcasting

Buat matriks 4×3 berisi angka 1-12. Lalu buat vektor `[100, 200, 300]` dan tambahkan ke matriks (broadcasting). Cetak matriks asli dan matriks hasil. Amati: vektor 3-elemen otomatis ditambahkan ke setiap baris matriks 4×3.

**Referensi singkat untuk Soal 7**:

```python
import numpy as np
mat = np.arange(1, 13).reshape(4, 3)
vec = np.array([100, 200, 300])
print("Matriks asli:")
print(mat)
print("\nMatriks + vektor:")
print(mat + vec)
```

Broadcasting: NumPy secara otomatis menyamakan shape dengan menyiarkan array kecil. Vektor panjang-3 di-broadcast ke bentuk (4, 3) sehingga bisa ditambahkan ke setiap baris matriks.

## Soal 8: DataFrame Sederhana

Buat DataFrame Pandas dengan 5 baris data mahasiswa (kolom: `nama`, `jurusan`, `ipk`). Data bebas, tapi harus realistis. Cetak DataFrame-nya, lalu cetak juga `shape`, `columns`, dan `dtypes`.

**Referensi singkat untuk Soal 8**:

```python
import pandas as pd
data = {
    'nama'    : ['Andi', 'Budi', 'Citra', 'Dewi', 'Eka'],
    'jurusan' : ['IF', 'SI', 'IF', 'TK', 'IF'],
    'ipk'     : [3.8, 3.5, 3.9, 3.2, 3.7]
}
df = pd.DataFrame(data)
print(df)
print(f"\nShape : {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"\nDtypes:\n{df.dtypes}")
```

DataFrame dibuat dari dictionary: key jadi nama kolom, value jadi data kolom. Tipe data otomatis terdeteksi: `nama` dan `jurusan` jadi `object` (string), `ipk` jadi `float64`.

## Soal 9: Filtering dan Group By

Dari DataFrame pada Soal 8 (atau buat baru dengan data yang lebih variatif, misalnya 10 mahasiswa), lakukan: (a) filter mahasiswa dengan IPK >= 3.7, (b) hitung rata-rata IPK per jurusan dengan `groupby`, (c) cetak mahasiswa dengan IPK tertinggi.

**Referensi singkat untuk Soal 9**:

```python
import pandas as pd
data = {
    'nama'    : ['Andi', 'Budi', 'Citra', 'Dewi', 'Eka', 'Fajar', 'Gita'],
    'jurusan' : ['IF', 'SI', 'IF', 'TK', 'IF', 'SI', 'TK'],
    'ipk'     : [3.8, 3.5, 3.9, 3.2, 3.7, 3.6, 3.4]
}
df = pd.DataFrame(data)

print("=== IPK >= 3.7 ===")
print(df[df['ipk'] >= 3.7])

print("\n=== Rata-rata IPK per jurusan ===")
print(df.groupby('jurusan')['ipk'].mean())

print("\n=== IPK tertinggi ===")
print(df.loc[df['ipk'].idxmax()])
```

Tiga pola yang sangat sering dipakai: (1) boolean indexing `df[kondisi]`, (2) group by dengan agregasi, (3) `idxmax()` untuk dapat index nilai tertinggi, lalu `df.loc[idx]` untuk dapat baris.

## Soal 10: Mini EDA dari Nol

Tantangan akhir: bangun EDA lengkap tanpa mencontek dari notebook utama. Generate dataset 50 mahasiswa sintetis (nama, jurusan, semester, IPK, nilai_uts, nilai_uas). Lalu: (1) cetak `head` dan `describe`, (2) hitung korelasi antar kolom numerik, (3) buat scatter plot IPK vs nilai_uts dengan matplotlib, (4) cetak 2-3 insight dari data.

**Referensi singkat untuk Soal 10**:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(0)
data = {
    'nama'     : [f"Mhs_{i}" for i in range(50)],
    'jurusan'  : np.random.choice(['IF', 'SI', 'TK'], 50),
    'semester' : np.random.randint(1, 9, 50),
    'ipk'      : np.clip(np.random.normal(3.4, 0.4, 50), 2.0, 4.0).round(2),
    'nilai_uts': np.clip(np.random.normal(72, 12, 50), 40, 100).astype(int),
    'nilai_uas': np.clip(np.random.normal(75, 10, 50), 40, 100).astype(int),
}
df = pd.DataFrame(data)

print("=== head(3) ===")
print(df.head(3))

print("\n=== describe ===")
print(df.describe().round(2))

print("\n=== korelasi ===")
print(df[['ipk', 'nilai_uts', 'nilai_uas', 'semester']].corr().round(3))

plt.figure(figsize=(8, 5))
plt.scatter(df['ipk'], df['nilai_uts'], alpha=0.6, color='steelblue')
plt.title('IPK vs Nilai UTS', fontweight='bold')
plt.xlabel('IPK')
plt.ylabel('Nilai UTS')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\n=== Insight ===")
print(f"- Korelasi IPK-UTS: {df['ipk'].corr(df['nilai_uts']):.3f}")
print(f"- Rata-rata IPK: {df['ipk'].mean():.2f}")
print(f"- Jurusan paling banyak: {df['jurusan'].value_counts().idxmax()}")
```

Mini EDA dari nol: generate data, inspect, hitung korelasi, visualisasi, tarik insight. Pola ini adalah 80% dari pekerjaan data analyst sehari-hari. Kalau kamu bisa membangunnya tanpa mencontek, kamu sudah siap untuk chapter berikutnya.
