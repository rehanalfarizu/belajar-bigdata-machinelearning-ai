## Soal 2: Variabel dan Cetak (3 baris)**Target**: Buat variabel `nama` berisi `"Rehan"`, lalu cetak variabel itu.**Expected output**:```Rehan```

````python
# Ketik ulang kode target di sini
````

## Soal 4: f-string Formatting (3 baris)**Target**: Cetak `"Rehan berumur 25 tahun"` pakai f-string, di mana nama dan umur adalah variabel.**Expected output**:```Rehan berumur 25 tahun```

````python
# Ketik ulang kode target di sini
````

**Referensi (jangan dilihat dulu)**:```pythonnama = "Rehan"umur = 25print(f"{nama} berumur {umur} tahun")```**Yang harus kamu perhatikan**:- Prefix `f` sebelum tanda kutip pembuka — tanpa `f`, `{nama}` dianggap string biasa- Kurung kurawal `{}` untuk tempat variabel, bukan `[]` atau `()`

## Soal 5: List dan Iterasi (5 baris)**Target**: Buat list berisi 3 nama, lalu cetak satu per satu pakai for loop.**Expected output**:```apeljerukmangga```

````python
# Ketik ulang kode target di sini
````

**Referensi (jangan dilihat dulu)**:```pythonbuah = ["apel", "jeruk", "mangga"]for item in buah:    print(item)```**Yang harus kamu perhatikan**:- List pakai kurung siku `[]`, elemen dipisah koma- `for item in buah:` diakhiri titik dua `:`- `print(item)` di-indentasi 4 spasi — bukan 2, bukan tab- Setelah baris `print(item)`, tidak ada lagi kode di-indentasi — itu artinya loop selesai di sana

## Soal 6: Fungsi Sederhana (4 baris)**Target**: Buat fungsi `sapa` yang menerima parameter `nama` dan return `"Halo, {nama}!"`. Lalu panggil dengan `"Budi"` dan cetak hasilnya.**Expected output**:```Halo, Budi!```

````python
# Ketik ulang kode target di sini
````

**Referensi (jangan dilihat dulu)**:```pythondef sapa(nama):    return f"Halo, {nama}!"print(sapa("Budi"))```**Yang harus kamu perhatikan**:- `def` (bukan `function` atau `func`)- Tanda kurung setelah nama fungsi, dengan parameter di dalam- Titik dua di akhir baris `def`- `return` di-indentasi 4 spasi- Pemanggilan fungsi: `sapa("Budi")`, bukan `sapa "Budi"` (Python tidak pakai spasi untuk argumen)

## Soal 7: If/Else (6 baris)**Target**: Cek apakah sebuah angka genap atau ganjil, lalu cetak `"Genap"` atau `"Ganjil"`.**Expected output** (untuk `angka = 42`):```42 adalah Genap```

````python
# Ketik ulang kode target di sini
````

**Referensi (jangan dilihat dulu)**:```pythonangka = 42if angka % 2 == 0:    print(f"{angka} adalah Genap")else:    print(f"{angka} adalah Ganjil")```**Yang harus kamu perhatikan**:- Modulo `%` untuk cek sisa pembagian- Perbandingan `==` (dua sama dengan), bukan `=` (satu sama dengan, itu assignment)- `if` dan `else` diakhiri titik dua- Dua baris print di-indentasi masing-masing 4 spasi

## Soal 8: Dict dan Akses (6 baris)**Target**: Buat dict `profil` dengan kunci `nama` dan `kota`, lalu cetak nama dan kotanya pakai f-string.**Expected output**:```Rehan tinggal di Jakarta```

````python
# Ketik ulang kode target di sini
````

**Referensi (jangan dilihat dulu)**:```pythonprofil = {    "nama": "Rehan",    "kota": "Jakarta"}print(f"{profil['nama']} tinggal di {profil['kota']}")```**Yang harus kamu perhatikan**:- Dict pakai kurung kurawal `{}`- Setiap item adalah `"kunci": nilai` dipisah titik dua- Tanda kutip tunggal `'nama'` di dalam f-string (bukan `"nama"` — itu akan bentrok dengan kutip f-string pembuka)

## Soal 9: Try/Except (8 baris)**Target**: Minta input angka dari pengguna. Kalau input bukan angka, cetak `"Input tidak valid"`.**Expected output** (kalau input `"bukan_angka"`):```Input tidak valid```**Expected output** (kalau input `"42"`):```42```

````python
# Ketik ulang kode target di sini
````

**Referensi (jangan dilihat dulu)**:```pythontry:    angka = int(input("Masukkan angka: "))    print(angka)except ValueError:    print("Input tidak valid")```**Yang harus kamu perhatikan**:- `try:` dan `except ValueError:` keduanya diakhiri titik dua- Blok `try` di-indentasi 4 spasi, blok `except` juga- `int(input(...))` di dalam `try` — kalau `input` bukan angka, `int()` memunculkan `ValueError` yang kita tangkap

## Soal 10: Program Lengkap (15 baris)**Target**: Buat program yang minta nama dan umur pengguna, lalu cetak sapaan. Tangani error kalau input umur bukan angka.**Expected output** (input nama=`"Ani"`, umur=`"20"`):```Halo Ani, umur 20 tahun```**Expected output** (input nama=`"Ani"`, umur=`"dua puluh"`):```Input umur harus angka```

````python
# Ketik ulang kode target di sini
````

**Referensi (jangan dilihat dulu)**:```pythonnama = input("Nama: ")try:    umur = int(input("Umur: "))    print(f"Halo {nama}, umur {umur} tahun")except ValueError:    print("Input umur harus angka")```**Yang harus kamu perhatikan**:- Baris pertama tidak perlu `try` — nama selalu string- `int(input(...))` di dalam `try` karena berpotensi error- f-string untuk pesan sukses, plain string untuk pesan error- Indentasi konsisten: dua baris di dalam `try` harus sejajar, satu baris di dalam `except` juga sejajar---**Selesai?** Kalau semua 10 soal bisa diketik ulang tanpa error, kamu sudah membangun 'memori otot' yang cukup kuat untuk sintaks Python dasar.
